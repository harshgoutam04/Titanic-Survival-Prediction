```python
from tensorflow.keras.models import load_model
import pickle 
import pandas as pd
import tensorflow as tf
```

```python
with open('label_encoder.pkl', 'rb') as file:
    label = pickle.load(file)
```

```python
with open('onehot_encoder.pkl', 'rb') as file:
    oht = pickle.load(file)
```

```python
with open('scaler.pkl', 'rb') as file:
    scaler = pickle.load(file)
```

```python
model=load_model('model.h5')
```

```python
user={'Pclass': 1, 'Sex': 'male', 'SibSp': 1, 'Parch': 0, 'Fare': 52, 'Embarked': 'Queenstown'}
```

```python
user=pd.DataFrame([user])
```

```python
user
```

```python
user['Sex']=label.transform(user['Sex'])
```

```python
embarked=oht.transform(user[['Embarked']])
```

```python
embarked=pd.DataFrame(embarked, columns=oht.get_feature_names_out())
```

```python
user=pd.concat([user, embarked], axis=1).drop('Embarked', axis=1)
```

```python
user
```

```python
user[['Pclass','SibSp','Parch','Fare']]=scaler.transform(user[['Pclass','SibSp','Parch','Fare']])
```

```python
user
```

```python
y=model.predict(user)
```
