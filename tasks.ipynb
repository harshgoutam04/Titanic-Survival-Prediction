# FEATURE ENCODING


```python
from sklearn.preprocessing import LabelEncoder
```

```python
label = LabelEncoder()
```

```python
df['Sex']=label.fit_transform(df['Sex'])
```

```python
from sklearn.preprocessing import OneHotEncoder
```

```python
oht=OneHotEncoder(sparse_output=False)
```

```python
Embarked=oht.fit_transform(df[['Embarked']])
```

```python
Embarked=pd.DataFrame(Embarked, columns=oht.get_feature_names_out())
```

```python
Embarked
```

```python
df=pd.concat([df.drop(columns=['Embarked']), Embarked], axis=1)
```

# TRAIN TEST SPLIT

```python
X=df.drop(columns=['Survived'])
y=df['Survived']
```

```python
X.isnull().sum()
```

```python
X=X.dropna()
```

```python
y.isnull().sum()
```

```python
y=y.dropna()
```

```python
df.isnull().sum()
```

```python
from sklearn.model_selection import train_test_split
```

```python
y = y.loc[X.index]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=20, random_state=42)
```

```python
X_train,X_val,y_train,y_val = train_test_split(X_train, y_train, test_size=0.2, random_state=42)
```

# MODEL BUILDING

```python
import tensorflow
```

```python
X_train.shape[1]
```

```python
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.callbacks import EarlyStopping
```

```python
model=Sequential([Dense(128,input_shape=(X_train.shape[1],), activation='relu'),
                  Dense(64, activation='relu'),
                  Dense(32, activation='relu'),
                Dense(1, activation='sigmoid')])
```

```python
model.summary()
```

```python
opt=tensorflow.keras.optimizers.Adam(learning_rate=0.01)
```

```python
model.compile(optimizer=opt, loss='binary_crossentropy', metrics=['accuracy'])
```

```python
EarlyStopping(monitor='val_loss', patience=5, restore_best_weights=True)
```

```python
model.fit(X_train, y_train, validation_data=(X_val, y_val), epochs=10)
```

```python
model.save('model.h5')
```
